# 1D CNN — Stock Price Prediction
Predicts next-close price from a sliding window of OHLCV bars using a 1D convolutional neural network (PyTorch).

**Workflow:** Fetch → Store (DB + CSV) → Load → Preprocess → Train → Evaluate

## 1. Imports

```python
import os
from datetime import date

import httpx
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split

print(f"PyTorch {torch.__version__} | device: {'cuda' if torch.cuda.is_available() else 'cpu'}")
```

## 2. Config

```python
# --- Data ---
SYMBOL      = "TSLA"
TIMEFRAME   = "1Min"
START_DATE  = "2017-01-01"
END_DATE    = str(date.today())
DATA_DIR    = os.path.join("..", "..", "data")   # project root /data
API_BASE    = "http://localhost:8000"

# --- Model ---
SEQ_LEN     = 30      # number of bars per input window
BATCH_SIZE  = 32
EPOCHS      = 50
LR          = 1e-3
TEST_SPLIT  = 0.2

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Symbol={SYMBOL}  Timeframe={TIMEFRAME}  {START_DATE} → {END_DATE}")
print("Using device:", DEVICE)
```

## 3. Fetch Data from Alpaca API
Calls the local `alpaca_api` FastAPI service (must be running: `uv run main.py`).
Fetches TSLA 1-minute bars and saves to both DB and CSV.

```python
# Single fetch — saves to both DB and CSV in one request
params = {
    "symbols": SYMBOL,
    "timeframe": TIMEFRAME,
    "start": START_DATE,
    "end": END_DATE,
    "save_to": "csv", # db,csv
}

with httpx.Client(timeout=None) as client:
    r = client.get(f"{API_BASE}/bars", params=params)
    r.raise_for_status()
    result = r.json()

bars = result.get("data", {}).get("bars", {}).get(SYMBOL, [])
print(f"Fetched {len(bars)} bars for {SYMBOL}")
print("Saved:", result.get("saved"))
```

## 4. Load Data

```python
csv_path = os.path.join(DATA_DIR, SYMBOL, f"{TIMEFRAME}.csv")
df = pd.read_csv(csv_path, parse_dates=["timestamp"])
df = df.sort_values("timestamp").reset_index(drop=True)

print(f"Loaded {len(df)} bars from {df['timestamp'].min()} to {df['timestamp'].max()}")
df[["timestamp", "open", "high", "low", "close", "volume"]].tail()
```

## 5. Preprocess

```python
FEATURES = ["open", "high", "low", "close", "volume"]
TARGET   = "close"

scaler = MinMaxScaler()
scaled = scaler.fit_transform(df[FEATURES])

# Build sliding-window sequences: X shape (N, SEQ_LEN, features), y shape (N,)
close_idx = FEATURES.index(TARGET)
X, y = [], []
for i in range(SEQ_LEN, len(scaled)):
    X.append(scaled[i - SEQ_LEN:i])
    y.append(scaled[i, close_idx])

X = np.array(X, dtype=np.float32)   # (N, SEQ_LEN, n_features)
y = np.array(y, dtype=np.float32)   # (N,)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=TEST_SPLIT, shuffle=False)
print(f"Train: {X_train.shape}  Test: {X_test.shape}")
```

## 6. Dataset & DataLoader

```python
class BarDataset(Dataset):
    def __init__(self, X, y):
        # Conv1d expects (batch, channels, length) — transpose seq and feature dims
        self.X = torch.tensor(X).permute(0, 2, 1)  # (N, n_features, SEQ_LEN)
        self.y = torch.tensor(y).unsqueeze(1)       # (N, 1)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]

train_loader = DataLoader(BarDataset(X_train, y_train), batch_size=BATCH_SIZE, shuffle=True)
test_loader  = DataLoader(BarDataset(X_test,  y_test),  batch_size=BATCH_SIZE)
```

## 7. Model

```python
class CNN1D(nn.Module):
    def __init__(self, n_features, seq_len):
        super().__init__()
        self.conv = nn.Sequential(
            nn.Conv1d(n_features, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.Conv1d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.AdaptiveAvgPool1d(1),  # global average pool → (batch, 128, 1)
        )
        self.fc = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, 1),
        )

    def forward(self, x):
        return self.fc(self.conv(x))

model = CNN1D(n_features=len(FEATURES), seq_len=SEQ_LEN).to(DEVICE)
print(model)
```

## 8. Train

```python
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=LR)

train_losses, val_losses = [], []

for epoch in range(1, EPOCHS + 1):
    model.train()
    epoch_loss = 0.0
    for X_batch, y_batch in train_loader:
        X_batch, y_batch = X_batch.to(DEVICE), y_batch.to(DEVICE)
        optimizer.zero_grad()
        loss = criterion(model(X_batch), y_batch)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
    train_losses.append(epoch_loss / len(train_loader))

    model.eval()
    with torch.no_grad():
        val_loss = sum(criterion(model(X_b.to(DEVICE)), y_b.to(DEVICE)).item()
                       for X_b, y_b in test_loader) / len(test_loader)
    val_losses.append(val_loss)

    if epoch % 10 == 0:
        print(f"Epoch {epoch:>3}/{EPOCHS}  train={train_losses[-1]:.5f}  val={val_losses[-1]:.5f}")
```

## 9. Evaluate

```python
# Loss curves
plt.figure(figsize=(10, 3))
plt.plot(train_losses, label="train")
plt.plot(val_losses,   label="val")
plt.xlabel("Epoch"); plt.ylabel("MSE"); plt.legend(); plt.title("Loss")
plt.tight_layout(); plt.show()

# Predicted vs actual (test set)
model.eval()
preds, actuals = [], []
with torch.no_grad():
    for X_batch, y_batch in test_loader:
        preds.append(model(X_batch.to(DEVICE)).cpu().numpy())
        actuals.append(y_batch.numpy())

preds   = np.concatenate(preds).flatten()
actuals = np.concatenate(actuals).flatten()

plt.figure(figsize=(12, 4))
plt.plot(actuals, label="Actual",    alpha=0.7)
plt.plot(preds,   label="Predicted", alpha=0.7)
plt.xlabel("Bar"); plt.ylabel("Scaled close"); plt.legend()
plt.title(f"{SYMBOL} — Predicted vs Actual Close (scaled)")
plt.tight_layout(); plt.show()
```
